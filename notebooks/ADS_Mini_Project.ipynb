```python
# ============================================================
# PART 5.1 — 95% CONFIDENCE INTERVAL FOR MEAN SALES
# ============================================================

from scipy import stats
import numpy as np

sales = df['Sales'].dropna()

n = len(sales)
mean_sales = sales.mean()
std_sales = sales.std()

confidence_level = 0.95
alpha = 1 - confidence_level

# t-critical value
t_critical = stats.t.ppf(1 - alpha/2, df=n-1)

# Standard error
standard_error = std_sales / np.sqrt(n)

# Margin of error
margin_error = t_critical * standard_error

# Confidence interval
lower_ci = mean_sales - margin_error
upper_ci = mean_sales + margin_error

print("95% Confidence Interval for Mean Sales")
print("-" * 50)
print(f"Sample Size       : {n}")
print(f"Mean Sales        : {mean_sales:.4f}")
print(f"Standard Error    : {standard_error:.4f}")
print(f"Margin of Error   : {margin_error:.4f}")
print(f"Lower Limit       : {lower_ci:.4f}")
print(f"Upper Limit       : {upper_ci:.4f}")
print("-" * 50)
```

```python
# ============================================================
# PART 5.2 — INDEPENDENT T-TEST
# Consumer vs Corporate
# ============================================================

consumer_sales = df.loc[
    df['Segment'] == 'Consumer',
    'Sales'
].dropna()

corporate_sales = df.loc[
    df['Segment'] == 'Corporate',
    'Sales'
].dropna()

print("Consumer observations :", len(consumer_sales))
print("Corporate observations:", len(corporate_sales))

print("\nMean Sales")
print("Consumer :", consumer_sales.mean())
print("Corporate:", corporate_sales.mean())
```

```python
# Welch's Independent t-test

t_stat, p_value = stats.ttest_ind(
    consumer_sales,
    corporate_sales,
    equal_var=False
)

alpha = 0.05

print("Independent t-test Results")
print("-" * 50)
print(f"t-statistic : {t_stat:.4f}")
print(f"p-value     : {p_value:.6f}")
print(f"Alpha       : {alpha}")

if p_value < alpha:
    print("\nDecision: Reject H0")
    print("Conclusion: There is a statistically significant difference "
          "in mean Sales between Consumer and Corporate customers.")
else:
    print("\nDecision: Fail to Reject H0")
    print("Conclusion: There is not enough statistical evidence "
          "of a difference in mean Sales between Consumer and Corporate customers.")
```

```python
# ============================================================
# PART 5.3 — ONE-WAY ANOVA
# Profit across Product Categories
# ============================================================

furniture_profit = df.loc[
    df['Category'] == 'Furniture',
    'Profit'
].dropna()

office_profit = df.loc[
    df['Category'] == 'Office Supplies',
    'Profit'
].dropna()

technology_profit = df.loc[
    df['Category'] == 'Technology',
    'Profit'
].dropna()

print("Number of observations:")
print("Furniture       :", len(furniture_profit))
print("Office Supplies :", len(office_profit))
print("Technology      :", len(technology_profit))

print("\nMean Profit:")
print("Furniture       :", furniture_profit.mean())
print("Office Supplies :", office_profit.mean())
print("Technology      :", technology_profit.mean())
```

```python
# One-way ANOVA

f_stat, p_value_anova = stats.f_oneway(
    furniture_profit,
    office_profit,
    technology_profit
)

alpha = 0.05

print("One-Way ANOVA Results")
print("-" * 50)
print(f"F-statistic : {f_stat:.4f}")
print(f"p-value     : {p_value_anova:.6f}")
print(f"Alpha       : {alpha}")

if p_value_anova < alpha:
    print("\nDecision: Reject H0")
    print("Conclusion: There is a statistically significant "
          "difference in mean Profit among the product categories.")
else:
    print("\nDecision: Fail to Reject H0")
    print("Conclusion: There is not enough statistical evidence "
          "that mean Profit differs among the product categories.")
```

```python
# ============================================================
# PART 5.4 — CHI-SQUARE TEST OF INDEPENDENCE
# Segment vs Category
# ============================================================

contingency_table = pd.crosstab(
    df['Segment'],
    df['Category']
)

print("Contingency Table")
display(contingency_table)
```

```python
chi2_stat, chi2_p, dof, expected = stats.chi2_contingency(
    contingency_table
)

alpha = 0.05

print("Chi-Square Test Results")
print("-" * 50)
print(f"Chi-square statistic : {chi2_stat:.4f}")
print(f"Degrees of freedom  : {dof}")
print(f"p-value             : {chi2_p:.6f}")
print(f"Alpha               : {alpha}")

if chi2_p < alpha:
    print("\nDecision: Reject H0")
    print("Conclusion: Customer Segment and Product Category "
          "are statistically associated.")
else:
    print("\nDecision: Fail to Reject H0")
    print("Conclusion: There is not enough statistical evidence "
          "of an association between Customer Segment and Product Category.")
```

```python
# ============================================================
# PART 7 — ANOMALY DETECTION
# METHOD 1: IQR
# ============================================================

def iqr_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    mask = (
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    )

    return mask, lower_bound, upper_bound


sales_iqr_mask, sales_lower, sales_upper = iqr_outliers(
    df, 'Sales'
)

profit_iqr_mask, profit_lower, profit_upper = iqr_outliers(
    df, 'Profit'
)

print("IQR Anomaly Detection")
print("-" * 50)

print(f"Sales lower bound : {sales_lower:.2f}")
print(f"Sales upper bound : {sales_upper:.2f}")
print(f"Sales anomalies   : {sales_iqr_mask.sum()}")

print()

print(f"Profit lower bound: {profit_lower:.2f}")
print(f"Profit upper bound: {profit_upper:.2f}")
print(f"Profit anomalies  : {profit_iqr_mask.sum()}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(x=df['Sales'], ax=axes[0])
axes[0].set_title('Sales — IQR Outlier Detection')
axes[0].set_xlabel('Sales')

sns.boxplot(x=df['Profit'], ax=axes[1])
axes[1].set_title('Profit — IQR Outlier Detection')
axes[1].set_xlabel('Profit')

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# METHOD 2: Z-SCORE
# ============================================================

from scipy.stats import zscore

df_anomaly = df.copy()

df_anomaly['Sales_ZScore'] = zscore(df_anomaly['Sales'])
df_anomaly['Profit_ZScore'] = zscore(df_anomaly['Profit'])

sales_z_mask = df_anomaly['Sales_ZScore'].abs() > 3
profit_z_mask = df_anomaly['Profit_ZScore'].abs() > 3

print("Z-score Anomaly Detection")
print("-" * 50)

print("Sales anomalies :", sales_z_mask.sum())
print("Profit anomalies:", profit_z_mask.sum())
```

```python
anomaly_comparison = pd.DataFrame({
    'Variable': ['Sales', 'Profit'],
    'IQR Anomalies': [
        sales_iqr_mask.sum(),
        profit_iqr_mask.sum()
    ],
    'Z-score Anomalies': [
        sales_z_mask.sum(),
        profit_z_mask.sum()
    ]
})

display(anomaly_comparison)
```

```python
# ============================================================
# PART 7.3 — ISOLATION FOREST
# ============================================================

from sklearn.ensemble import IsolationForest

# Select numerical transaction features
anomaly_features = [
    'Sales',
    'Quantity',
    'Discount',
    'Profit'
]

X_anomaly = df[anomaly_features].copy()

# Isolation Forest
iso_forest = IsolationForest(
    n_estimators=200,
    contamination='auto',
    random_state=42
)

iso_predictions = iso_forest.fit_predict(X_anomaly)

# -1 = anomaly, 1 = normal
df_anomaly['IsolationForest'] = iso_predictions

# Count anomalies
isolation_anomalies = (iso_predictions == -1).sum()
isolation_normal = (iso_predictions == 1).sum()

print("Isolation Forest Results")
print("-" * 50)
print("Normal observations   :", isolation_normal)
print("Anomalous observations:", isolation_anomalies)
print("Total observations    :", len(df_anomaly))
```

```python
# ============================================================
# PART 7.4 — COMPARISON OF ALL ANOMALY DETECTION METHODS
# ============================================================

anomaly_comparison_final = pd.DataFrame({
    'Variable / Method': [
        'Sales - IQR',
        'Sales - Z-score',
        'Profit - IQR',
        'Profit - Z-score',
        'Isolation Forest'
    ],
    'Anomalies': [
        sales_iqr_mask.sum(),
        sales_z_mask.sum(),
        profit_iqr_mask.sum(),
        profit_z_mask.sum(),
        isolation_anomalies
    ]
})

anomaly_comparison_final['Percentage'] = (
    anomaly_comparison_final['Anomalies']
    / len(df) * 100
)

display(
    anomaly_comparison_final.round(2)
)
```
